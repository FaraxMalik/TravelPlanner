## 3. Check Dataset Size and Memory Usage

```python
# Memory usage analysis function
def analyze_memory_usage(df):
    """
    Analyze and display memory usage of the dataframe
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("MEMORY USAGE ANALYSIS")
    print("=" * 50)
    
    # Basic info
    print(f"📊 Dataset Shape: {df.shape}")
    print(f"📊 Total Rows: {df.shape[0]:,}")
    print(f"📊 Total Columns: {df.shape[1]:,}")
    
    # Memory usage
    memory_usage = df.memory_usage(deep=True)
    total_memory = memory_usage.sum()
    
    print(f"\n💾 Total Memory Usage: {total_memory / 1024**2:.2f} MB")
    print(f"💾 Average Memory per Row: {total_memory / len(df):.2f} bytes")
    
    # Memory usage by column
    print(f"\n🔍 Memory Usage by Column (Top 10):")
    memory_by_col = memory_usage.sort_values(ascending=False).head(10)
    for col, mem in memory_by_col.items():
        if col != 'Index':
            print(f"   {col}: {mem / 1024**2:.2f} MB")
    
    # Data type analysis
    print(f"\n📋 Data Types Summary:")
    dtype_counts = df.dtypes.value_counts()
    for dtype, count in dtype_counts.items():
        print(f"   {dtype}: {count} columns")
    
    return total_memory

# Function to optimize data types
def optimize_dtypes(df):
    """
    Optimize data types to reduce memory usage
    """
    if df is None:
        return None
    
    print("\n🔧 OPTIMIZING DATA TYPES...")
    original_memory = df.memory_usage(deep=True).sum()
    
    for col in df.columns:
        col_type = df[col].dtype
        
        if col_type != 'object':
            c_min = df[col].min()
            c_max = df[col].max()
            
            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                    
            elif str(col_type)[:5] == 'float':
                if c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
    
    new_memory = df.memory_usage(deep=True).sum()
    reduction = (original_memory - new_memory) / original_memory * 100
    
    print(f"✅ Memory usage reduced by {reduction:.1f}%")
    print(f"💾 New memory usage: {new_memory / 1024**2:.2f} MB")
    
    return df

# Analyze current dataset
if 'df_sample' in locals() and df_sample is not None:
    analyze_memory_usage(df_sample)
    
    # Optimize data types for the sample
    df_sample_optimized = optimize_dtypes(df_sample.copy())
else:
    print("⚠️  Please load the dataset first by running the previous cell.")
```

## 4. Examine Data Types and Structure

```python
def analyze_data_structure(df):
    """
    Comprehensive analysis of data types and structure
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("DATA STRUCTURE ANALYSIS")
    print("=" * 50)
    
    # Basic info
    print(f"📊 Dataset Info:")
    df.info(memory_usage='deep')
    
    print(f"\n📋 Detailed Column Analysis:")
    print("-" * 80)
    
    structure_analysis = []
    
    for col in df.columns:
        col_info = {
            'Column': col,
            'Data_Type': str(df[col].dtype),
            'Non_Null_Count': df[col].count(),
            'Null_Count': df[col].isnull().sum(),
            'Null_Percentage': round((df[col].isnull().sum() / len(df)) * 100, 2),
            'Unique_Values': df[col].nunique(),
            'Memory_Usage_MB': round(df[col].memory_usage(deep=True) / 1024**2, 3)
        }
        
        # Add sample values for better understanding
        if df[col].dtype == 'object':
            col_info['Sample_Values'] = list(df[col].dropna().unique()[:5])
            col_info['Is_Categorical'] = 'Likely' if df[col].nunique() < len(df) * 0.05 else 'Unlikely'
        else:
            col_info['Min_Value'] = df[col].min() if df[col].count() > 0 else 'N/A'
            col_info['Max_Value'] = df[col].max() if df[col].count() > 0 else 'N/A'
            col_info['Mean_Value'] = round(df[col].mean(), 2) if df[col].count() > 0 else 'N/A'
        
        structure_analysis.append(col_info)
    
    # Convert to DataFrame for better display
    structure_df = pd.DataFrame(structure_analysis)
    
    print("📊 Column Summary:")
    display(structure_df[['Column', 'Data_Type', 'Non_Null_Count', 'Null_Percentage', 
                         'Unique_Values', 'Memory_Usage_MB']])
    
    # Categorize columns
    print(f"\n🏷️  Column Categories:")
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
    datetime_cols = df.select_dtypes(include=['datetime64']).columns.tolist()
    
    print(f"   📈 Numerical columns ({len(numerical_cols)}): {numerical_cols[:10]}")
    if len(numerical_cols) > 10:
        print(f"      ... and {len(numerical_cols) - 10} more")
    
    print(f"   🏷️  Categorical columns ({len(categorical_cols)}): {categorical_cols[:10]}")
    if len(categorical_cols) > 10:
        print(f"      ... and {len(categorical_cols) - 10} more")
    
    if datetime_cols:
        print(f"   📅 DateTime columns ({len(datetime_cols)}): {datetime_cols}")
    
    # Identify potential issues
    print(f"\n⚠️  Potential Data Type Issues:")
    issues_found = False
    
    for col in categorical_cols:
        if df[col].nunique() > len(df) * 0.9:
            print(f"   • '{col}': High cardinality ({df[col].nunique()} unique values) - might be an ID column")
            issues_found = True
    
    for col in numerical_cols:
        if df[col].dtype == 'float64' and df[col].isnull().sum() == 0:
            if all(df[col] == df[col].astype(int)):
                print(f"   • '{col}': Contains only integers but stored as float")
                issues_found = True
    
    if not issues_found:
        print("   ✅ No obvious data type issues found")
    
    return structure_df

# Analyze structure
if 'df_sample' in locals() and df_sample is not None:
    structure_summary = analyze_data_structure(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

## 5. Check for Missing Values

```python
def analyze_missing_values(df):
    """
    Comprehensive missing values analysis with visualizations
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("MISSING VALUES ANALYSIS")
    print("=" * 50)
    
    # Calculate missing values
    missing_data = df.isnull().sum()
    missing_percent = (missing_data / len(df)) * 100
    
    # Create missing data summary
    missing_summary = pd.DataFrame({
        'Column': missing_data.index,
        'Missing_Count': missing_data.values,
        'Missing_Percentage': missing_percent.values
    }).sort_values('Missing_Percentage', ascending=False)
    
    # Filter columns with missing values
    columns_with_missing = missing_summary[missing_summary['Missing_Count'] > 0]
    
    print(f"📊 Missing Values Summary:")
    print(f"   Total columns: {len(df.columns)}")
    print(f"   Columns with missing values: {len(columns_with_missing)}")
    print(f"   Columns without missing values: {len(df.columns) - len(columns_with_missing)}")
    
    if len(columns_with_missing) > 0:
        print(f"\n🔍 Columns with Missing Values:")
        display(columns_with_missing.head(20))
        
        # Visualize missing values
        plt.figure(figsize=(15, 8))
        
        # Plot 1: Missing values bar chart
        plt.subplot(2, 2, 1)
        top_missing = columns_with_missing.head(15)
        plt.barh(range(len(top_missing)), top_missing['Missing_Percentage'])
        plt.yticks(range(len(top_missing)), top_missing['Column'])
        plt.xlabel('Missing Percentage (%)')
        plt.title('Top 15 Columns with Missing Values')
        plt.gca().invert_yaxis()
        
        # Plot 2: Missing values heatmap (if reasonable number of columns)
        if len(df.columns) <= 50:
            plt.subplot(2, 2, 2)
            # Select subset for heatmap visualization
            missing_matrix = df.isnull()
            sns.heatmap(missing_matrix.iloc[:100], cbar=True, yticklabels=False)
            plt.title('Missing Values Pattern (First 100 rows)')
            plt.xlabel('Columns')
        
        # Plot 3: Missing values distribution
        plt.subplot(2, 2, 3)
        missing_counts = missing_summary['Missing_Percentage']
        plt.hist(missing_counts, bins=20, edgecolor='black')
        plt.xlabel('Missing Percentage (%)')
        plt.ylabel('Number of Columns')
        plt.title('Distribution of Missing Values')
        
        # Plot 4: Completeness overview
        plt.subplot(2, 2, 4)
        complete_data = 100 - missing_percent.mean()
        plt.pie([complete_data, 100-complete_data], 
                labels=[f'Complete\n{complete_data:.1f}%', f'Missing\n{100-complete_data:.1f}%'],
                autopct='%1.1f%%', startangle=90)
        plt.title('Overall Data Completeness')
        
        plt.tight_layout()
        plt.show()
        
        # Missing value patterns analysis
        print(f"\n🔍 Missing Value Patterns:")
        
        # Check for rows with high missing values
        row_missing_count = df.isnull().sum(axis=1)
        rows_with_many_missing = (row_missing_count > len(df.columns) * 0.5).sum()
        
        print(f"   • Rows with >50% missing values: {rows_with_many_missing}")
        print(f"   • Maximum missing values in a single row: {row_missing_count.max()}")
        print(f"   • Average missing values per row: {row_missing_count.mean():.2f}")
        
        # Check for completely empty columns
        completely_empty = (missing_data == len(df)).sum()
        if completely_empty > 0:
            print(f"   ⚠️  Completely empty columns: {completely_empty}")
        
        # Recommendations
        print(f"\n💡 Recommendations:")
        high_missing = columns_with_missing[columns_with_missing['Missing_Percentage'] > 50]
        if len(high_missing) > 0:
            print(f"   • Consider dropping columns with >50% missing: {list(high_missing['Column'][:5])}")
        
        moderate_missing = columns_with_missing[
            (columns_with_missing['Missing_Percentage'] > 10) & 
            (columns_with_missing['Missing_Percentage'] <= 50)
        ]
        if len(moderate_missing) > 0:
            print(f"   • Consider imputation for columns with 10-50% missing: {list(moderate_missing['Column'][:5])}")
        
    else:
        print("✅ No missing values found in the dataset!")
    
    return missing_summary

# Analyze missing values
if 'df_sample' in locals() and df_sample is not None:
    missing_analysis = analyze_missing_values(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

## 6. Identify Duplicate Rows

```python
def analyze_duplicates(df):
    """
    Comprehensive duplicate analysis
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("DUPLICATE ANALYSIS")
    print("=" * 50)
    
    # Overall duplicate analysis
    total_rows = len(df)
    duplicate_rows = df.duplicated().sum()
    duplicate_percentage = (duplicate_rows / total_rows) * 100
    
    print(f"📊 Overall Duplicate Summary:")
    print(f"   Total rows: {total_rows:,}")
    print(f"   Duplicate rows: {duplicate_rows:,}")
    print(f"   Duplicate percentage: {duplicate_percentage:.2f}%")
    print(f"   Unique rows: {total_rows - duplicate_rows:,}")
    
    if duplicate_rows > 0:
        print(f"\n🔍 Duplicate Rows Analysis:")
        
        # Show some duplicate examples
        duplicated_mask = df.duplicated(keep=False)
        duplicate_examples = df[duplicated_mask].head(10)
        
        if len(duplicate_examples) > 0:
            print("Sample duplicate rows:")
            display(duplicate_examples)
        
        # Analyze duplicates by specific columns (if reasonable number of columns)
        if len(df.columns) <= 20:
            print(f"\n🔍 Column-wise Duplicate Analysis:")
            for col in df.columns:
                col_duplicates = df[col].duplicated().sum()
                col_duplicate_pct = (col_duplicates / total_rows) * 100
                unique_values = df[col].nunique()
                
                if col_duplicate_pct > 0:
                    print(f"   {col}: {col_duplicates:,} duplicates ({col_duplicate_pct:.2f}%), "
                          f"{unique_values:,} unique values")
        
        # Visualize duplicates
        plt.figure(figsize=(12, 6))
        
        # Plot 1: Duplicate vs Unique rows
        plt.subplot(1, 2, 1)
        labels = ['Unique Rows', 'Duplicate Rows']
        sizes = [total_rows - duplicate_rows, duplicate_rows]
        colors = ['lightblue', 'lightcoral']
        plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=90)
        plt.title('Duplicate vs Unique Rows')
        
        # Plot 2: Duplicate patterns (if not too many columns)
        if len(df.columns) <= 15:
            plt.subplot(1, 2, 2)
            col_duplicates = []
            col_names = []
            for col in df.columns[:10]:  # Top 10 columns
                dup_count = df[col].duplicated().sum()
                if dup_count > 0:
                    col_duplicates.append(dup_count)
                    col_names.append(col)
            
            if col_duplicates:
                plt.barh(range(len(col_names)), col_duplicates)
                plt.yticks(range(len(col_names)), col_names)
                plt.xlabel('Number of Duplicates')
                plt.title('Duplicates by Column (Top 10)')
                plt.gca().invert_yaxis()
        
        plt.tight_layout()
        plt.show()
        
        # Recommendations
        print(f"\n💡 Recommendations:")
        if duplicate_percentage > 10:
            print(f"   ⚠️  High duplicate percentage ({duplicate_percentage:.1f}%) - consider removing duplicates")
        elif duplicate_percentage > 1:
            print(f"   • Moderate duplicates ({duplicate_percentage:.1f}%) - investigate and consider removal")
        else:
            print(f"   • Low duplicate percentage ({duplicate_percentage:.1f}%) - minimal impact")
        
        print(f"   • To remove duplicates: df.drop_duplicates()")
        print(f"   • To keep last occurrence: df.drop_duplicates(keep='last')")
        
    else:
        print("✅ No duplicate rows found!")
    
    # Check for near-duplicates (if dataset is not too large)
    if len(df) <= 10000:  # Only for smaller samples
        print(f"\n🔍 Near-Duplicate Analysis (excluding exact matches):")
        # This is a simplified near-duplicate check
        # In practice, you might want more sophisticated methods
        
        text_columns = df.select_dtypes(include=['object']).columns
        if len(text_columns) > 0:
            print(f"   Checking text columns for similar values...")
            for col in text_columns[:3]:  # Check first 3 text columns
                if df[col].nunique() < len(df) * 0.9:  # If reasonable cardinality
                    value_counts = df[col].value_counts()
                    similar_threshold = 3  # Values appearing 3+ times
                    frequent_values = value_counts[value_counts >= similar_threshold]
                    if len(frequent_values) > 0:
                        print(f"     '{col}': {len(frequent_values)} values appear {similar_threshold}+ times")

# Analyze duplicates
if 'df_sample' in locals() and df_sample is not None:
    analyze_duplicates(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

## 7. Analyze Data Distribution

```python
def analyze_distributions(df):
    """
    Comprehensive data distribution analysis
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("DATA DISTRIBUTION ANALYSIS")
    print("=" * 50)
    
    # Separate numerical and categorical columns
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
    
    print(f"📊 Distribution Overview:")
    print(f"   Numerical columns: {len(numerical_cols)}")
    print(f"   Categorical columns: {len(categorical_cols)}")
    
    # Numerical distributions
    if len(numerical_cols) > 0:
        print(f"\n📈 NUMERICAL DISTRIBUTIONS")
        print("-" * 40)
        
        # Statistical summary
        print("Statistical Summary:")
        display(df[numerical_cols].describe())
        
        # Plot distributions for numerical columns
        n_cols = min(len(numerical_cols), 12)  # Limit to 12 columns for visualization
        cols_to_plot = numerical_cols[:n_cols]
        
        if len(cols_to_plot) > 0:
            fig, axes = plt.subplots(3, 4, figsize=(20, 15))
            axes = axes.ravel() if len(cols_to_plot) > 1 else [axes]
            
            for i, col in enumerate(cols_to_plot):
                if i < len(axes):
                    # Histogram
                    df[col].hist(bins=30, ax=axes[i], alpha=0.7, edgecolor='black')
                    axes[i].set_title(f'{col}\nMean: {df[col].mean():.2f}, Std: {df[col].std():.2f}')
                    axes[i].set_xlabel(col)
                    axes[i].set_ylabel('Frequency')
            
            # Hide empty subplots
            for i in range(len(cols_to_plot), len(axes)):
                axes[i].set_visible(False)
            
            plt.suptitle('Distribution of Numerical Variables', fontsize=16)
            plt.tight_layout()
            plt.show()
        
        # Box plots for outlier detection
        if len(cols_to_plot) > 0:
            fig, axes = plt.subplots(2, 2, figsize=(15, 12))
            axes = axes.ravel()
            
            for i, col in enumerate(cols_to_plot[:4]):  # Top 4 columns
                df.boxplot(column=col, ax=axes[i])
                axes[i].set_title(f'Box Plot: {col}')
            
            # Hide empty subplots
            for i in range(len(cols_to_plot[:4]), len(axes)):
                axes[i].set_visible(False)
            
            plt.suptitle('Box Plots for Outlier Detection', fontsize=16)
            plt.tight_layout()
            plt.show()
        
        # Normality assessment
        print(f"\n📊 Distribution Shape Analysis:")
        for col in numerical_cols[:10]:  # Analyze top 10 numerical columns
            skewness = df[col].skew()
            kurtosis = df[col].kurtosis()
            
            # Interpret skewness
            if abs(skewness) < 0.5:
                skew_interpretation = "Approximately symmetric"
            elif abs(skewness) < 1:
                skew_interpretation = "Moderately skewed"
            else:
                skew_interpretation = "Highly skewed"
            
            print(f"   {col}:")
            print(f"     Skewness: {skewness:.3f} ({skew_interpretation})")
            print(f"     Kurtosis: {kurtosis:.3f}")
    
    # Categorical distributions
    if len(categorical_cols) > 0:
        print(f"\n🏷️  CATEGORICAL DISTRIBUTIONS")
        print("-" * 40)
        
        for col in categorical_cols[:10]:  # Analyze top 10 categorical columns
            print(f"\n📊 {col}:")
            value_counts = df[col].value_counts()
            
            print(f"   Unique values: {df[col].nunique()}")
            print(f"   Most frequent: '{value_counts.index[0]}' ({value_counts.iloc[0]} times, {(value_counts.iloc[0]/len(df)*100):.1f}%)")
            
            if df[col].nunique() <= 20:  # Show distribution for low cardinality
                print(f"   Top 10 values:")
                for val, count in value_counts.head(10).items():
                    pct = (count / len(df)) * 100
                    print(f"     '{val}': {count} ({pct:.1f}%)")
            else:
                print(f"   High cardinality column - showing top 5:")
                for val, count in value_counts.head(5).items():
                    pct = (count / len(df)) * 100
                    print(f"     '{val}': {count} ({pct:.1f}%)")
        
        # Visualize categorical distributions
        categorical_to_plot = [col for col in categorical_cols[:6] if df[col].nunique() <= 20]
        
        if len(categorical_to_plot) > 0:
            fig, axes = plt.subplots(2, 3, figsize=(18, 12))
            axes = axes.ravel() if len(categorical_to_plot) > 1 else [axes]
            
            for i, col in enumerate(categorical_to_plot):
                if i < len(axes):
                    value_counts = df[col].value_counts().head(15)
                    value_counts.plot(kind='bar', ax=axes[i])
                    axes[i].set_title(f'{col} Distribution')
                    axes[i].set_xlabel(col)
                    axes[i].set_ylabel('Count')
                    axes[i].tick_params(axis='x', rotation=45)
            
            # Hide empty subplots
            for i in range(len(categorical_to_plot), len(axes)):
                axes[i].set_visible(False)
            
            plt.suptitle('Distribution of Categorical Variables', fontsize=16)
            plt.tight_layout()
            plt.show()
    
    # Correlation analysis for numerical variables
    if len(numerical_cols) > 1:
        print(f"\n🔗 CORRELATION ANALYSIS")
        print("-" * 40)
        
        correlation_matrix = df[numerical_cols].corr()
        
        # Find highly correlated pairs
        high_corr_pairs = []
        for i in range(len(correlation_matrix.columns)):
            for j in range(i+1, len(correlation_matrix.columns)):
                corr_val = correlation_matrix.iloc[i, j]
                if abs(corr_val) > 0.7:  # High correlation threshold
                    high_corr_pairs.append((
                        correlation_matrix.columns[i], 
                        correlation_matrix.columns[j], 
                        corr_val
                    ))
        
        if high_corr_pairs:
            print("High correlations found (|r| > 0.7):")
            for col1, col2, corr in high_corr_pairs:
                print(f"   {col1} ↔ {col2}: {corr:.3f}")
        else:
            print("No high correlations found (|r| > 0.7)")
        
        # Correlation heatmap
        if len(numerical_cols) <= 20:  # Only plot if reasonable number of variables
            plt.figure(figsize=(12, 10))
            sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
                       square=True, fmt='.2f')
            plt.title('Correlation Matrix of Numerical Variables')
            plt.tight_layout()
            plt.show()

# Analyze distributions
if 'df_sample' in locals() and df_sample is not None:
    analyze_distributions(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

## 8. Check for Outliers

```python
def detect_outliers(df):
    """
    Comprehensive outlier detection using multiple methods
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("OUTLIER DETECTION ANALYSIS")
    print("=" * 50)
    
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    
    if len(numerical_cols) == 0:
        print("❌ No numerical columns found for outlier analysis.")
        return
    
    print(f"📊 Analyzing {len(numerical_cols)} numerical columns for outliers")
    
    outlier_summary = []
    
    for col in numerical_cols:
        if df[col].count() == 0:  # Skip columns with no data
            continue
            
        # Method 1: IQR method
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        iqr_outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
        iqr_outlier_count = len(iqr_outliers)
        iqr_outlier_pct = (iqr_outlier_count / len(df)) * 100
        
        # Method 2: Z-score method (for normally distributed data)
        z_scores = np.abs((df[col] - df[col].mean()) / df[col].std())
        z_outliers = df[z_scores > 3][col]  # 3 standard deviations
        z_outlier_count = len(z_outliers)
        z_outlier_pct = (z_outlier_count / len(df)) * 100
        
        # Method 3: Modified Z-score (more robust)
        median = df[col].median()
        mad = np.median(np.abs(df[col] - median))
        modified_z_scores = 0.6745 * (df[col] - median) / mad if mad != 0 else np.zeros(len(df))
        modified_z_outliers = df[np.abs(modified_z_scores) > 3.5][col]
        modified_z_outlier_count = len(modified_z_outliers)
        modified_z_outlier_pct = (modified_z_outlier_count / len(df)) * 100
        
        outlier_info = {
            'Column': col,
            'IQR_Outliers': iqr_outlier_count,
            'IQR_Percentage': round(iqr_outlier_pct, 2),
            'Z_Score_Outliers': z_outlier_count,
            'Z_Score_Percentage': round(z_outlier_pct, 2),
            'Modified_Z_Outliers': modified_z_outlier_count,
            'Modified_Z_Percentage': round(modified_z_outlier_pct, 2),
            'Min_Value': df[col].min(),
            'Max_Value': df[col].max(),
            'Mean': round(df[col].mean(), 2),
            'Std': round(df[col].std(), 2)
        }
        
        outlier_summary.append(outlier_info)
    
    # Create summary DataFrame
    outlier_df = pd.DataFrame(outlier_summary)
    
    print(f"\n📊 Outlier Detection Summary:")
    display(outlier_df[['Column', 'IQR_Outliers', 'IQR_Percentage', 
                      'Z_Score_Outliers', 'Z_Score_Percentage']])
    
    # Visualize outliers for top columns with outliers
    high_outlier_cols = outlier_df[outlier_df['IQR_Percentage'] > 1]['Column'].tolist()[:6]
    
    if len(high_outlier_cols) > 0:
        print(f"\n📊 Visualizing outliers for columns with >1% outliers:")
        
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        axes = axes.ravel() if len(high_outlier_cols) > 1 else [axes]
        
        for i, col in enumerate(high_outlier_cols):
            if i < len(axes):
                # Box plot
                df.boxplot(column=col, ax=axes[i])
                axes[i].set_title(f'{col}\nIQR Outliers: {outlier_df[outlier_df["Column"]==col]["IQR_Outliers"].iloc[0]}')
                axes[i].set_ylabel('Values')
        
        # Hide empty subplots
        for i in range(len(high_outlier_cols), len(axes)):
            axes[i].set_visible(False)
        
        plt.suptitle('Box Plots for Columns with High Outlier Counts', fontsize=16)
        plt.tight_layout()
        plt.show()
        
        # Scatter plots to show outlier patterns
        if len(high_outlier_cols) >= 2:
            fig, axes = plt.subplots(1, 2, figsize=(15, 6))
            
            # Plot 1: First two columns with high outliers
            col1, col2 = high_outlier_cols[0], high_outlier_cols[1]
            axes[0].scatter(df[col1], df[col2], alpha=0.6)
            axes[0].set_xlabel(col1)
            axes[0].set_ylabel(col2)
            axes[0].set_title(f'Scatter Plot: {col1} vs {col2}')
            
            # Plot 2: Outlier distribution
            outlier_counts = outlier_df['IQR_Outliers'].head(10)
            col_names = outlier_df['Column'].head(10)
            axes[1].barh(range(len(col_names)), outlier_counts)
            axes[1].set_yticks(range(len(col_names)))
            axes[1].set_yticklabels(col_names)
            axes[1].set_xlabel('Number of Outliers (IQR Method)')
            axes[1].set_title('Outlier Count by Column')
            axes[1].invert_yaxis()
            
            plt.tight_layout()
            plt.show()
    
    # Detailed analysis for most problematic columns
    most_outliers = outlier_df.nlargest(3, 'IQR_Percentage')
    
    if len(most_outliers) > 0:
        print(f"\n🔍 Detailed Analysis of Columns with Most Outliers:")
        
        for _, row in most_outliers.iterrows():
            col = row['Column']
            print(f"\n📈 {col}:")
            print(f"   IQR outliers: {row['IQR_Outliers']} ({row['IQR_Percentage']}%)")
            print(f"   Z-score outliers: {row['Z_Score_Outliers']} ({row['Z_Score_Percentage']}%)")
            print(f"   Value range: {row['Min_Value']} to {row['Max_Value']}")
            print(f"   Mean ± Std: {row['Mean']} ± {row['Std']}")
            
            # Show extreme values
            extreme_high = df[col].nlargest(5)
            extreme_low = df[col].nsmallest(5)
            
            print(f"   Top 5 highest values: {list(extreme_high.values)}")
            print(f"   Top 5 lowest values: {list(extreme_low.values)}")
    
    # Recommendations
    print(f"\n💡 Outlier Treatment Recommendations:")
    
    severe_outliers = outlier_df[outlier_df['IQR_Percentage'] > 10]
    moderate_outliers = outlier_df[(outlier_df['IQR_Percentage'] > 2) & (outlier_df['IQR_Percentage'] <= 10)]
    mild_outliers = outlier_df[(outlier_df['IQR_Percentage'] > 0) & (outlier_df['IQR_Percentage'] <= 2)]
    
    if len(severe_outliers) > 0:
        print(f"   🚨 Severe outliers (>10%): {list(severe_outliers['Column'])}")
        print(f"      → Consider transformation, capping, or investigation for data errors")
    
    if len(moderate_outliers) > 0:
        print(f"   ⚠️  Moderate outliers (2-10%): {list(moderate_outliers['Column'])}")
        print(f"      → Consider winsorization or robust scaling")
    
    if len(mild_outliers) > 0:
        print(f"   ℹ️  Mild outliers (<2%): {list(mild_outliers['Column'])}")
        print(f"      → May be legitimate extreme values, monitor during modeling")
    
    no_outliers = outlier_df[outlier_df['IQR_Percentage'] == 0]
    if len(no_outliers) > 0:
        print(f"   ✅ No outliers detected: {list(no_outliers['Column'])}")
    
    return outlier_df

# Detect outliers
if 'df_sample' in locals() and df_sample is not None:
    outlier_analysis = detect_outliers(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

## 9. Validate Data Consistency

```python
def validate_data_consistency(df):
    """
    Comprehensive data consistency validation
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 50)
    print("DATA CONSISTENCY VALIDATION")
    print("=" * 50)
    
    consistency_issues = []
    
    # 1. Check for negative values in columns that shouldn't have them
    print("🔍 Checking for logical inconsistencies...")
    
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    
    # Common columns that shouldn't be negative
    potential_positive_cols = [col for col in numerical_cols 
                              if any(keyword in col.lower() for keyword in 
                                   ['age', 'price', 'cost', 'amount', 'count', 'quantity', 
                                    'size', 'length', 'width', 'height', 'distance', 'duration'])]
    
    for col in potential_positive_cols:
        negative_count = (df[col] < 0).sum()
        if negative_count > 0:
            consistency_issues.append({
                'Type': 'Negative Values',
                'Column': col,
                'Issue': f'{negative_count} negative values in typically positive column',
                'Severity': 'High'
            })
            print(f"   ⚠️  {col}: {negative_count} negative values")
    
    # 2. Check for impossible date ranges
    date_cols = df.select_dtypes(include=['datetime64']).columns.tolist()
    potential_date_cols = [col for col in df.columns 
                          if any(keyword in col.lower() for keyword in 
                               ['date', 'time', 'year', 'month', 'day'])]
    
    for col in potential_date_cols:
        if df[col].dtype == 'object':
            # Try to identify date-like strings
            sample_values = df[col].dropna().head(100)
            date_like_count = 0
            for val in sample_values:
                if isinstance(val, str) and any(char in val for char in ['/', '-', ':']):
                    date_like_count += 1
            
            if date_like_count > 50:  # More than 50% look like dates
                consistency_issues.append({
                    'Type': 'Date Format',
                    'Column': col,
                    'Issue': 'Potential date column stored as text',
                    'Severity': 'Medium'
                })
                print(f"   📅 {col}: Potential date column stored as text")
    
    # 3. Check for inconsistent text formatting
    text_cols = df.select_dtypes(include=['object']).columns.tolist()
    
    for col in text_cols[:10]:  # Check first 10 text columns
        if df[col].nunique() > 1:
            # Check for mixed case issues
            sample_values = df[col].dropna().astype(str).head(1000)
            
            # Count different case patterns
            uppercase_count = sum(1 for val in sample_values if val.isupper())
            lowercase_count = sum(1 for val in sample_values if val.islower())
            mixed_case_count = len(sample_values) - uppercase_count - lowercase_count
            
            if uppercase_count > 0 and lowercase_count > 0 and mixed_case_count > 0:
                consistency_issues.append({
                    'Type': 'Text Formatting',
                    'Column': col,
                    'Issue': f'Mixed case formatting (Upper: {uppercase_count}, Lower: {lowercase_count}, Mixed: {mixed_case_count})',
                    'Severity': 'Low'
                })
            
            # Check for leading/trailing spaces
            spaces_count = sum(1 for val in sample_values if val != val.strip())
            if spaces_count > 0:
                consistency_issues.append({
                    'Type': 'Text Formatting',
                    'Column': col,
                    'Issue': f'{spaces_count} values with leading/trailing spaces',
                    'Severity': 'Medium'
                })
    
    # 4. Check for unrealistic ranges
    print(f"\n🔍 Checking for unrealistic value ranges...")
    
    for col in numerical_cols:
        col_min = df[col].min()
        col_max = df[col].max()
        col_mean = df[col].mean()
        col_std = df[col].std()
        
        # Check for extremely wide ranges (might indicate data entry errors)
        if col_std > 0:
            range_ratio = (col_max - col_min) / col_std
            if range_ratio > 100:  # Very wide range compared to standard deviation
                consistency_issues.append({
                    'Type': 'Wide Range',
                    'Column': col,
                    'Issue': f'Extremely wide range: {col_min} to {col_max} (ratio: {range_ratio:.1f})',
                    'Severity': 'Medium'
                })
        
        # Common sense checks for specific column types
        if 'age' in col.lower():
            if col_min < 0 or col_max > 150:
                consistency_issues.append({
                    'Type': 'Unrealistic Values',
                    'Column': col,
                    'Issue': f'Unrealistic age range: {col_min} to {col_max}',
                    'Severity': 'High'
                })
        
        elif 'year' in col.lower():
            current_year = datetime.now().year
            if col_min < 1900 or col_max > current_year + 10:
                consistency_issues.append({
                    'Type': 'Unrealistic Values',
                    'Column': col,
                    'Issue': f'Unrealistic year range: {col_min} to {col_max}',
                    'Severity': 'High'
                })
    
    # 5. Check for ID columns that might have duplicates
    print(f"\n🔍 Checking ID columns for consistency...")
    
    potential_id_cols = [col for col in df.columns 
                        if any(keyword in col.lower() for keyword in 
                             ['id', 'key', 'index', 'code', 'number']) 
                        and df[col].dtype in ['int64', 'object']]
    
    for col in potential_id_cols:
        duplicate_count = df[col].duplicated().sum()
        if duplicate_count > 0:
            consistency_issues.append({
                'Type': 'ID Duplicates',
                'Column': col,
                'Issue': f'{duplicate_count} duplicate values in potential ID column',
                'Severity': 'High'
            })
    
    # 6. Check for placeholder values
    print(f"\n🔍 Checking for placeholder/default values...")
    
    common_placeholders = ['unknown', 'null', 'n/a', 'na', 'none', 'undefined', 
                          'missing', 'tbd', 'tba', '999', '-999', '0', 'test']
    
    for col in text_cols:
        for placeholder in common_placeholders:
            placeholder_count = df[col].str.lower().eq(placeholder).sum()
            if placeholder_count > len(df) * 0.05:  # More than 5% of data
                consistency_issues.append({
                    'Type': 'Placeholder Values',
                    'Column': col,
                    'Issue': f'{placeholder_count} instances of placeholder "{placeholder}"',
                    'Severity': 'Medium'
                })
    
    # Display results
    if consistency_issues:
        print(f"\n⚠️  CONSISTENCY ISSUES FOUND ({len(consistency_issues)} issues):")
        
        issues_df = pd.DataFrame(consistency_issues)
        
        # Group by severity
        high_severity = issues_df[issues_df['Severity'] == 'High']
        medium_severity = issues_df[issues_df['Severity'] == 'Medium']
        low_severity = issues_df[issues_df['Severity'] == 'Low']
        
        if len(high_severity) > 0:
            print(f"\n🚨 HIGH SEVERITY ISSUES ({len(high_severity)}):")
            display(high_severity)
        
        if len(medium_severity) > 0:
            print(f"\n⚠️  MEDIUM SEVERITY ISSUES ({len(medium_severity)}):")
            display(medium_severity)
        
        if len(low_severity) > 0:
            print(f"\nℹ️  LOW SEVERITY ISSUES ({len(low_severity)}):")
            display(low_severity)
        
        # Visualize issues by type
        issue_type_counts = issues_df['Type'].value_counts()
        
        plt.figure(figsize=(12, 6))
        
        plt.subplot(1, 2, 1)
        issue_type_counts.plot(kind='bar')
        plt.title('Consistency Issues by Type')
        plt.xlabel('Issue Type')
        plt.ylabel('Count')
        plt.xticks(rotation=45)
        
        plt.subplot(1, 2, 2)
        severity_counts = issues_df['Severity'].value_counts()
        severity_counts.plot(kind='pie', autopct='%1.1f%%')
        plt.title('Issues by Severity')
        
        plt.tight_layout()
        plt.show()
        
    else:
        print("✅ No major consistency issues found!")
    
    # Data quality score
    total_checks = len(numerical_cols) + len(text_cols) + len(potential_id_cols)
    issues_score = max(0, 100 - (len(consistency_issues) / max(total_checks, 1)) * 100)
    
    print(f"\n📊 Data Consistency Score: {issues_score:.1f}/100")
    
    if issues_score >= 90:
        print("✅ Excellent data consistency!")
    elif issues_score >= 75:
        print("👍 Good data consistency with minor issues")
    elif issues_score >= 50:
        print("⚠️  Moderate data consistency - several issues need attention")
    else:
        print("🚨 Poor data consistency - major cleanup required")
    
    return consistency_issues

# Validate data consistency
if 'df_sample' in locals() and df_sample is not None:
    consistency_results = validate_data_consistency(df_sample)
else:
    print("⚠️  Please load the dataset first.")
```

```python
def generate_data_quality_report(df):
    """
    Generate a comprehensive data quality report
    """
    if df is None:
        print("⚠️  No dataset loaded yet.")
        return
    
    print("=" * 60)
    print("📋 COMPREHENSIVE DATA QUALITY REPORT")
    print("=" * 60)
    
    # Basic dataset information
    print(f"\n📊 DATASET OVERVIEW")
    print("-" * 30)
    print(f"Dataset Shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
    print(f"Total Cells: {df.shape[0] * df.shape[1]:,}")
    print(f"Memory Usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    print(f"Analysis Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Data quality metrics
    total_cells = df.shape[0] * df.shape[1]
    missing_cells = df.isnull().sum().sum()
    duplicate_rows = df.duplicated().sum()
    
    completeness_score = ((total_cells - missing_cells) / total_cells) * 100
    uniqueness_score = ((df.shape[0] - duplicate_rows) / df.shape[0]) * 100
    
    print(f"\n📈 DATA QUALITY METRICS")
    print("-" * 30)
    print(f"Completeness: {completeness_score:.2f}% ({total_cells - missing_cells:,}/{total_cells:,} cells)")
    print(f"Uniqueness: {uniqueness_score:.2f}% ({df.shape[0] - duplicate_rows:,}/{df.shape[0]:,} rows)")
    
    # Column analysis summary
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
    datetime_cols = df.select_dtypes(include=['datetime64']).columns.tolist()
    
    print(f"\n🏷️  COLUMN TYPE DISTRIBUTION")
    print("-" * 30)
    print(f"Numerical columns: {len(numerical_cols)} ({len(numerical_cols)/df.shape[1]*100:.1f}%)")
    print(f"Categorical columns: {len(categorical_cols)} ({len(categorical_cols)/df.shape[1]*100:.1f}%)")
    print(f"DateTime columns: {len(datetime_cols)} ({len(datetime_cols)/df.shape[1]*100:.1f}%)")
    
    # Missing values summary
    missing_summary = df.isnull().sum()
    columns_with_missing = (missing_summary > 0).sum()
    
    print(f"\n❌ MISSING VALUES SUMMARY")
    print("-" * 30)
    print(f"Columns with missing values: {columns_with_missing}/{df.shape[1]} ({columns_with_missing/df.shape[1]*100:.1f}%)")
    print(f"Total missing cells: {missing_cells:,}")
    print(f"Missing percentage: {(missing_cells/total_cells)*100:.2f}%")
    
    if columns_with_missing > 0:
        worst_missing = missing_summary[missing_summary > 0].nlargest(5)
        print(f"\nTop 5 columns with most missing values:")
        for col, count in worst_missing.items():
            pct = (count / len(df)) * 100
            print(f"  • {col}: {count:,} ({pct:.1f}%)")
    
    # Duplicates summary
    print(f"\n🔄 DUPLICATES SUMMARY")
    print("-" * 30)
    print(f"Duplicate rows: {duplicate_rows:,} ({(duplicate_rows/df.shape[0])*100:.2f}%)")
    print(f"Unique rows: {df.shape[0] - duplicate_rows:,}")
    
    # Data types and potential issues
    print(f"\n🔧 DATA TYPE RECOMMENDATIONS")
    print("-" * 30)
    
    recommendations = []
    
    # Check for potential optimization
    for col in numerical_cols:
        if df[col].dtype == 'float64':
            if df[col].isnull().sum() == 0 and all(df[col] == df[col].astype(int)):
                recommendations.append(f"Convert '{col}' from float64 to int64 (contains only integers)")
        
        if df[col].dtype == 'int64':
            col_min, col_max = df[col].min(), df[col].max()
            if col_min >= 0 and col_max <= 255:
                recommendations.append(f"Convert '{col}' to uint8 for memory efficiency")
            elif col_min >= -128 and col_max <= 127:
                recommendations.append(f"Convert '{col}' to int8 for memory efficiency")
    
    # Check categorical columns
    for col in categorical_cols:
        unique_ratio = df[col].nunique() / len(df)
        if unique_ratio < 0.05:  # Less than 5% unique values
            recommendations.append(f"Convert '{col}' to categorical type for memory efficiency")
        elif unique_ratio > 0.95:  # More than 95% unique values
            recommendations.append(f"'{col}' has high cardinality - consider if it's an identifier")
    
    if recommendations:
        print("Optimization suggestions:")
        for i, rec in enumerate(recommendations[:10], 1):
            print(f"  {i}. {rec}")
    else:
        print("✅ No immediate optimization recommendations")
    
    # Overall data quality score
    print(f"\n🎯 OVERALL DATA QUALITY ASSESSMENT")
    print("-" * 30)
    
    # Calculate weighted scores
    weights = {
        'completeness': 0.3,
        'uniqueness': 0.2,
        'consistency': 0.2,
        'validity': 0.3
    }
    
    # Consistency score (simplified - based on data type appropriateness)
    consistency_score = 85  # Base score
    if len(recommendations) > 5:
        consistency_score -= 10
    
    # Validity score (based on realistic ranges and formats)
    validity_score = 90  # Base score
    for col in numerical_cols:
        if 'age' in col.lower():
            if df[col].min() < 0 or df[col].max() > 150:
                validity_score -= 10
                break
    
    overall_score = (
        completeness_score * weights['completeness'] +
        uniqueness_score * weights['uniqueness'] +
        consistency_score * weights['consistency'] +
        validity_score * weights['validity']
    )
    
    print(f"Completeness Score: {completeness_score:.1f}/100 (weight: {weights['completeness']*100:.0f}%)")
    print(f"Uniqueness Score: {uniqueness_score:.1f}/100 (weight: {weights['uniqueness']*100:.0f}%)")
    print(f"Consistency Score: {consistency_score:.1f}/100 (weight: {weights['consistency']*100:.0f}%)")
    print(f"Validity Score: {validity_score:.1f}/100 (weight: {weights['validity']*100:.0f}%)")
    print(f"\n🏆 OVERALL QUALITY SCORE: {overall_score:.1f}/100")
    
    # Quality assessment
    if overall_score >= 90:
        quality_level = "🟢 EXCELLENT"
        assessment = "Dataset is high quality and ready for ML modeling with minimal preprocessing."
    elif overall_score >= 75:
        quality_level = "🟡 GOOD"
        assessment = "Dataset quality is good with some areas for improvement."
    elif overall_score >= 60:
        quality_level = "🟠 FAIR"
        assessment = "Dataset requires moderate cleaning and preprocessing before ML modeling."
    else:
        quality_level = "🔴 POOR"
        assessment = "Dataset requires significant cleaning and preprocessing."
    
    print(f"\nQuality Level: {quality_level}")
    print(f"Assessment: {assessment}")
    
    # Action items
    print(f"\n✅ RECOMMENDED ACTIONS")
    print("-" * 30)
    
    action_items = []
    
    if missing_cells > total_cells * 0.05:  # More than 5% missing
        action_items.append("Address missing values through imputation or removal")
    
    if duplicate_rows > df.shape[0] * 0.01:  # More than 1% duplicates
        action_items.append("Remove or investigate duplicate rows")
    
    if len(recommendations) > 0:
        action_items.append("Optimize data types for better memory efficiency")
    
    if overall_score < 75:
        action_items.append("Conduct detailed data cleaning and validation")
    
    action_items.append("Consider feature engineering based on business requirements")
    action_items.append("Validate data splits for ML modeling (train/validation/test)")
    
    for i, action in enumerate(action_items, 1):
        print(f"{i}. {action}")
    
    # Summary for next steps
    print(f"\n🚀 NEXT STEPS FOR ML MODELING")
    print("-" * 30)
    print("1. Load the full dataset (currently analyzing sample)")
    print("2. Apply identified data cleaning steps")
    print("3. Engineer features based on domain knowledge")
    print("4. Split data into train/validation/test sets")
    print("5. Scale/normalize features as needed")
    print("6. Begin model selection and training")
    
    print(f"\n" + "=" * 60)
    print("📋 DATA QUALITY REPORT COMPLETE")
    print("=" * 60)
    
    # Return summary for programmatic use
    return {
        'overall_score': overall_score,
        'completeness_score': completeness_score,
        'uniqueness_score': uniqueness_score,
        'consistency_score': consistency_score,
        'validity_score': validity_score,
        'missing_cells': missing_cells,
        'duplicate_rows': duplicate_rows,
        'recommendations': recommendations,
        'action_items': action_items,
        'quality_level': quality_level
    }

# Generate the comprehensive report
if 'df_sample' in locals() and df_sample is not None:
    print("🔄 Generating comprehensive data quality report...")
    quality_report = generate_data_quality_report(df_sample)
else:
    print("⚠️  Please load the dataset first by running the previous cells.")
```
